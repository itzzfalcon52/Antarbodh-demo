###SST Dataset Idea
so,the sst dataset has the shape of (1,150,200) in time,lat,long and the sea_surface_temperature as the variable for the dataset.The whole dataset is in netCDF format,so we will have to use xarrays.

Dimensions:these are the actuals values(or rows as we say in pandas) of the dataset, which are time, latitude, and longitude. The time dimension has 1 value, latitude has 150 values, and longitude has 200 values.

Shape: The shape of the dataset is (1, 150, 200), which corresponds to the dimensions of time, latitude, and longitude respectively. This means that for each time point (in this case, just one), there are 150 latitude points and 200 longitude points, resulting in a grid of sea surface temperature values.

coordinates: The coordinates of the dataset are the actual values of time, latitude, and longitude. The time coordinate will have a single value (e.g., a specific date), while the latitude and longitude coordinates will have 150 and 200 values respectively, representing the spatial grid of the sea surface temperature data.

data variables: The main data variable in the SST dataset is the sea_surface_temperature, which contains the temperature values at each grid point defined by the latitude and longitude coordinates. This variable will have a shape of (1, 150, 200), corresponding to the dimensions of time, latitude, and longitude.So,it is basically the output value that we are interested in analyzing or visualizing.

attributes: The attributes of the dataset provide additional metadata about the data, such as units, descriptions, and any other relevant information. For example, the sea_surface_temperature variable may have an attribute indicating that the temperature is measured in degrees Celsius. Other attributes may include information about the source of the data, the time range covered by the dataset, and any processing steps that were applied to the data.
we can also find attributes for a particular data variable, such as the sea_surface_temperature, which may include details about the measurement method, calibration, or any quality control procedures that were applied to ensure the accuracy of the data. These attributes are important for understanding the context and reliability of the dataset, and they can help guide further analysis or interpretation of the sea surface temperature values.
